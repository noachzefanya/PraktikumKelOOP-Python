```python
from abc import ABC, abstractmethod

# 1. Abstraction (Kerangka Dasar) & Encapsulation (Keamanan Data)
class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = 0
        self.__harga_dasar = harga_dasar

    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar.split(' ')[0]} {self.nama_kamar.split(' ')[1]}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar.split(' ')[0]} {self.nama_kamar.split(' ')[1]}: {jumlah} unit.")

    @abstractmethod
    def tampilkan_detail(self, jumlah_malam):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass
```

```python
# 2. Class Kamar Deluxe (Inheritance & Polymorphism)
class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        self.fasilitas = fasilitas

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 0.10
        return (harga + pajak) * jumlah_malam

    def tampilkan_detail(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 0.10
        subtotal = self.hitung_harga_total(jumlah_malam)

        fmt_harga = f"{int(harga):,}".replace(',', '.')
        fmt_pajak = f"{int(pajak):,}".replace(',', '.')
        fmt_sub = f"{int(subtotal):,}".replace(',', '.')

        return (f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}\n"
                f" Harga Dasar/Malam: Rp {fmt_harga} | Pajak(10%): Rp {fmt_pajak}\n"
                f" Menginap: {jumlah_malam} malam | Subtotal: Rp {fmt_sub}")
```

```python
# 3. Class Kamar Standard (Inheritance & Polymorphism)
class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        self.kapasitas = kapasitas

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 0.05
        return (harga + pajak) * jumlah_malam

    def tampilkan_detail(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * 0.05
        subtotal = self.hitung_harga_total(jumlah_malam)

        fmt_harga = f"{int(harga):,}".replace(',', '.')
        fmt_pajak = f"{int(pajak):,}".replace(',', '.')
        fmt_sub = f"{int(subtotal):,}".replace(',', '.')

        return (f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}\n"
                f" Harga Dasar/Malam: Rp {fmt_harga} | Pajak(5%): Rp {fmt_pajak}\n"
                f" Menginap: {jumlah_malam} malam | Subtotal: Rp {fmt_sub}")
```

```python
# 4. Fungsi Proses Transaksi
def proses_transaksi(daftar_pesanan):
    print("\n--- STRUK PEMESANAN ---")
    total_tagihan = 0
    for i, pesanan in enumerate(daftar_pesanan, 1):
        kamar = pesanan['kamar']
        malam = pesanan['malam']
        print(f"{i}. {kamar.tampilkan_detail(malam)}")
        total_tagihan += kamar.hitung_harga_total(malam)

    print("-" * 40)
    fmt_total = f"{int(total_tagihan):,}".replace(',', '.')
    print(f"TOTAL TAGIHAN: Rp {fmt_total}")
    print("-" * 40)
```

```python
# 5. Alur Program (User Story)
print("--- SETUP DATA KAMAR ---")

# a) Admin membuat data kamar
kamar1 = KamarDeluxe("Kamar Deluxe Sea View", 1500000, "Private Pool")
kamar2 = KamarStandard("Kamar Standard Superior", 500000, "2 Orang")

# b) Admin mencoba mengisi stok kamar dengan angka negatif & positif
kamar1.tambah_stok(10)
kamar2.tambah_stok(-5)
kamar2.tambah_stok(20)

# c) Tamu memesan 2 malam Kamar Deluxe dan 1 malam Kamar Standard.
pesanan_tamu = [
    {'kamar': kamar1, 'malam': 2},
    {'kamar': kamar2, 'malam': 1}
]

# d) Menampilkan detail tagihan
proses_transaksi(pesanan_tamu)
```
